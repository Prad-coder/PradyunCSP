## Highlights from My Projects
- During N@TM in both 2nd trimester and 3rd trimester, I was thrilled to present my work in the Striver social media site and the Tower Defense game integrated with Open Coding Society
- Striver: I was able to get key feedback through a Google Form on our presentation and the actual site that was presented. It seemed like many liked the UI but people said that the features could be a little more complex.
- Tower Defense: I was able to showcase my Clash Royale themed tower defense game which seemed like a hit compared to other games at the event. I was able to get key feedback, such as a more clean UI as well as an easier UX through a more in depth help page and better instructions for the chaos in the game.

## Highlights from Other Projects
- During N@TM in 3rd trimester, I was able to visit a lot of project integrated with both Open Coding Society and Pilot City
- I was interested in the CrossWise project the most, which tracked border crossing times through present and historical data in weather, traffic, and the time of day
- I was also interested in the student_2026 project for Open Coding Society, which was a new and easier to use repository for next years' students to learn quicker and use easier
